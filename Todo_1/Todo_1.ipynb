```python
import matplotlib.pyplot as plt

def plot_simulation(t, x, u, title):
    """Helper to plot simulation results for analysis."""
    fig, axs = plt.subplots(3, 2, figsize=(12, 10))
    fig.suptitle(title)

    # Position
    axs[0, 0].plot(t, x[9, :], label='x')
    axs[0, 0].plot(t, x[10, :], label='y')
    axs[0, 0].plot(t, x[11, :], label='z')
    axs[0, 0].set_title('Position')
    axs[0, 0].legend()
    axs[0, 0].grid(True)

    # Velocity
    axs[1, 0].plot(t, x[6, :], label='vx')
    axs[1, 0].plot(t, x[7, :], label='vy')
    axs[1, 0].plot(t, x[8, :], label='vz')
    axs[1, 0].set_title('Velocity')
    axs[1, 0].legend()
    axs[1, 0].grid(True)

    # Euler Angles
    axs[0, 1].plot(t, np.rad2deg(x[3, :]), label='alpha (x)')
    axs[0, 1].plot(t, np.rad2deg(x[4, :]), label='beta (y)')
    axs[0, 1].plot(t, np.rad2deg(x[5, :]), label='gamma (z)')
    axs[0, 1].set_title('Euler Angles (deg)')
    axs[0, 1].legend()
    axs[0, 1].grid(True)

    # Angular Velocity
    axs[1, 1].plot(t, x[0, :], label='wx')
    axs[1, 1].plot(t, x[1, :], label='wy')
    axs[1, 1].plot(t, x[2, :], label='wz')
    axs[1, 1].set_title('Angular Velocity (rad/s)')
    axs[1, 1].legend()
    axs[1, 1].grid(True)

    # Inputs
    # Use t instead of t[:-1] to match u's dimension (both are N)
    axs[2, 0].step(t, np.rad2deg(u[0, :]), label='d1', where='post')
    axs[2, 0].step(t, np.rad2deg(u[1, :]), label='d2', where='post')
    axs[2, 0].set_title('Servo Inputs (deg)')
    axs[2, 0].legend()
    axs[2, 0].grid(True)

    axs[2, 1].step(t, u[2, :], label='Pavg', where='post')
    axs[2, 1].step(t, u[3, :], label='Pdiff', where='post')
    axs[2, 1].set_title('Throttle Inputs')
    axs[2, 1].legend()
    axs[2, 1].grid(True)

    plt.tight_layout()
    plt.show()

def analyze_simulation(t, x, title):
    """
    Prints a textual summary of the simulation for automated analysis.
    This allows the 'engineer' (AI) to 'see' the results without visual plots.
    """
    print(f"--- Analysis: {title} ---")
    print(f"Duration: {t[-1]:.2f}s")
    
    # Final State
    final_pos = x[9:12, -1]
    final_vel = x[6:9, -1]
    final_rpy = np.rad2deg(x[3:6, -1])
    
    print(f"Final Position (x,y,z): [{final_pos[0]:.4f}, {final_pos[1]:.4f}, {final_pos[2]:.4f}] m")
    print(f"Final Velocity (vx,vy,vz): [{final_vel[0]:.4f}, {final_vel[1]:.4f}, {final_vel[2]:.4f}] m/s")
    print(f"Final Attitude (r,p,y): [{final_rpy[0]:.4f}, {final_rpy[1]:.4f}, {final_rpy[2]:.4f}] deg")
    
    # Key Statistics
    max_z = np.max(x[11, :])
    min_z = np.min(x[11, :])
    print(f"Z-range: [{min_z:.4f}, {max_z:.4f}] m")
    
    # Check for movement in other axes
    x_range = np.ptp(x[9, :])
    y_range = np.ptp(x[10, :])
    print(f"Displacement X: {x_range:.4f} m, Y: {y_range:.4f} m")
    print("-" * 30 + "\n")

# Todo 1.2: Simulate various scenarios

# 1. Hover
print("Simulating Hover...")
x0 = np.zeros(12)
u_hover = np.array([0, 0, 66.67, 0]) # Approx hover
Tf = 2.0
t, x, u_hist = rocket.simulate(x0, Tf, u_hover, method='nonlinear')
plot_simulation(t, x, u_hist, "Hover Simulation")
analyze_simulation(t, x, "Hover Simulation")

# 2. Ascend
print("Simulating Ascent...")
u_ascend = np.array([0, 0, 70.0, 0]) # > 66.67
t, x, u_hist = rocket.simulate(x0, Tf, u_ascend, method='nonlinear')
plot_simulation(t, x, u_hist, "Ascent Simulation")
analyze_simulation(t, x, "Ascent Simulation")

# 3. Rotate about Z (Roll)
print("Simulating Roll (Gamma)...")
u_roll = np.array([0, 0, 66.67, 5.0]) # Pdiff > 0
t, x, u_hist = rocket.simulate(x0, Tf, u_roll, method='nonlinear')
plot_simulation(t, x, u_hist, "Roll Simulation")
analyze_simulation(t, x, "Roll Simulation")

# 4. Fly along X (requires Pitch/Beta)
# To fly X, we need to tilt. 
# sys_x uses delta2. Let's try delta2 > 0.
print("Simulating X-Movement (via Delta2)...")
u_move_x = np.array([0, np.deg2rad(5), 66.67, 0]) 
t, x, u_hist = rocket.simulate(x0, Tf, u_move_x, method='nonlinear')
plot_simulation(t, x, u_hist, "X-Movement Simulation")
analyze_simulation(t, x, "X-Movement Simulation")
```

    *   **Linear Dynamics:** Computes acceleration in the world frame using the rotation matrix $T_{wb}$ and gravity vector, $\dot{v} = T_{wb} {}^b F / m - g$.
